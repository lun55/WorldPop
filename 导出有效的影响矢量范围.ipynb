```python
import pandas as pd
import geopandas as gpd

# 1. 加载数据
VECTOR_DICT = {
    0: r"F:\机场噪音\GAUL_2024_调整\GAUL_2024_L0\GAUL_2024_L0.shp",
    2: r"F:\机场噪音\GAUL_2024_调整\GAUL_2024_L2\GAUL_2024_L2.shp"
}
L0Data_path = r"F:\机场噪音\Noise_Population_Mean\总\L0\SEL_oneday\L0_intersected_SEL_oneday_45dB.csv"
L2Data_path = r"F:\机场噪音\Noise_Population_Mean\总\L2\SEL_oneday\L2_intersected_SEL_oneday_45dB.csv"

# 读取 CSV (只保留必要的列以节省内存)
df_L0 = pd.read_csv(L0Data_path)
df_L2 = pd.read_csv(L2Data_path)

# 读取 SHP
gpd_L0 = gpd.read_file(VECTOR_DICT[0])
gpd_L2 = gpd.read_file(VECTOR_DICT[2])

# ---------------- 2. 处理 L0 (国家级) ----------------
# 挂接键通常是 gaul0_name
merged_L0 = gpd_L0.merge(df_L0[['gaul0_name', 'affected_pop']], on='gaul0_name', how='inner')
# 筛选受影响人口 > 0 的矢量
v_L0_affected = merged_L0[merged_L0['affected_pop'] > 0]

# ---------------- 3. 处理 L2 (市级/区级) ----------------
# L2 需要多字段匹配以确保唯一性
l2_keys = ['gaul0_name', 'gaul1_name', 'gaul2_name']
merged_L2 = gpd_L2.merge(df_L2[l2_keys + ['affected_pop']], on=l2_keys, how='inner')
# 筛选受影响人口 > 0 的矢量
v_L2_affected = merged_L2[merged_L2['affected_pop'] > 0]

# ---------------- 4. 保存结果 ----------------
v_L0_affected.to_file("F:\机场噪音\GAUL_2024_调整\影响区域\L0_Affected_Areas.shp", encoding='utf-8')
v_L2_affected.to_file("F:\机场噪音\GAUL_2024_调整\影响区域\L2_Affected_Areas.shp", encoding='utf-8')

print(f"L0 提取数量: {len(v_L0_affected)}")
print(f"L2 提取数量: {len(v_L2_affected)}")
```
